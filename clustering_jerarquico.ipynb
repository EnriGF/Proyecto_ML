```python
#importamos libreria y datos
import pandas as pd
url_datos = 'https://drive.google.com/uc?id=1OzVlp4RsT63W6d18_Zc5WmUonLXMDNL8'
mexpr = pd.read_csv(url_datos, delimiter='\t')
mexpr.head()        #Comprobar que los datos se han cargado correctamente
```

```python
#Librerias necesarias para el clustering jerárquico
import seaborn as sns
from matplotlib import pyplot as plt

#Representación clásica por heatmap. Vamos a coger pocas muestras para que sea visible
mexpr_toy = mexpr.iloc[0:49, 0:19]
sns.clustermap(mexpr_toy)
#Esto genera un heatmap en el que se conectan los 49 genes por el dendograma, se ve la expresión y las características.
#Además, resulta sencillo dividir el resultado en patrones, se pueden ver 4 patrones horizontales
```

```python
#En el heatmap se pueden modificar argumentos para cambiar por ejemplo la medida de distancia
#(Euclidean, correlation...) o el método (single, complete, centroid...)
#Por ejemplo:
sns.clustermap(mexpr_toy, metric="euclidean", method="centroid")
```
